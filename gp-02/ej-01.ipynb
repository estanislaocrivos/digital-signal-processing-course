```python
from matplotlib.pyplot import (
    figure,
    grid,
    plot,
    show,
    step,
    title,
    xlabel,
    xticks,
    ylabel,
)
from numpy import (
    arange,
    cos,
    linspace,
    pi,
)
from scipy.signal import freqz
```

```python
sampling_freq_hz = 30
duration_s = 3
n_samples = int(sampling_freq_hz * duration_s)

t = arange(n_samples) / sampling_freq_hz

x = 3 + 4 * cos(10 * pi * t) + 5 * cos(14 * pi * t) + 2 * cos(20 * pi * t)

figure(1, figsize=(8, 6), dpi=200)
step(t, x)
xlabel("Time [s]")
ylabel("Amplitude")
title("x[n]")
grid(True)
show()
```

```python
f = linspace(-80, 80, 8000 + 1)
f, H = freqz(x, worN=f, fs=sampling_freq_hz)

figure(1, figsize=(8, 6), dpi=200)
plot(f, abs(H))
xlabel("Frequency [Hz]")
ylabel("Magnitude")
xticks(arange(-80, 81, 10))
title("x[n] frequency spectrum")
grid(True)
show()
```

Para poder recuperar la señal original a partir del espectro de la señal muestreada, es necesario filtrar las componentes de frecuencia que se encuentran por encima de $f_s/2$, ya que el espectro de la señal digital se repite con esta tasa. Para esto, se puede utilizar un filtro pasa-bajos con una frecuencia de corte de $f_c = f_s/2 = 15Hz$.
