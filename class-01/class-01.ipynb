## Energy of a signal

A signal can be classified as either a power signal or an energy signal based on its total energy and average power. The energy of a continuous signal $x(t)$ is defined as:

$E_x = \int_{-\infty}^{\infty} |x(t)|^2 dt$

While the energy of a discrete signal $x[n]$ is defined as:

$E_x = \sum_{n=-\infty}^{\infty} |x[n]|^2$

```python
sampling_freq_hz = 10e3
duration_s = 1
n_samples = int(duration_s * sampling_freq_hz)

t = arange(n_samples) / sampling_freq_hz
t_ms = t * 1e3

signal_alpha = 10
signal_amplitude = 1
signal = signal_amplitude * exp(-signal_alpha * t)

signal_energy = sum(abs(signal) ** 2) * 1 / sampling_freq_hz
signal_energy_theoretical = abs(signal_amplitude) ** 2 / (2 * signal_alpha)
print(f"Signal energy = {signal_energy}")
print(f"Signal energy (theoretical) = {signal_energy_theoretical}")

figure(1, figsize=(8, 6), dpi=200)
plot(t_ms, signal)
xlabel("Time [ms]")
ylabel("Amplitude")
grid(True)
show()
```
